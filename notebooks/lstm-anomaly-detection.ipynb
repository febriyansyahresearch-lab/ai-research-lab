# LSTM — Anomaly Detection on Synthetic Log Data

Deteksi anomali pada log sintetis menggunakan LSTM untuk cybersecurity.

```python
import torch
import torch.nn as nn
import torch.optim as optim
from src.models.lstm import LSTMClassifier
from src.utils.data_loader import generate_synthetic_logs, get_dataloader
```

```python
X, y = generate_synthetic_logs(n_samples=500, seq_length=20)
loader = get_dataloader(X, y, batch_size=32)
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = LSTMClassifier(input_dim=4, hidden_dim=64, num_layers=2, num_classes=2).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

```python
for epoch in range(5):
    for seq, lbl in loader:
        seq, lbl = seq.to(device), lbl.to(device)
        optimizer.zero_grad()
        loss = criterion(model(seq), lbl)
        loss.backward()
        optimizer.step()
    print(f'Epoch {epoch+1}, Loss: {loss.item():.4f}')
```

```python
correct = total = 0
with torch.no_grad():
    for seq, lbl in loader:
        seq, lbl = seq.to(device), lbl.to(device)
        preds = model(seq).argmax(dim=1)
        correct += (preds == lbl).sum().item()
        total += lbl.size(0)
print(f'Accuracy: {100 * correct / total:.2f}%')
```
