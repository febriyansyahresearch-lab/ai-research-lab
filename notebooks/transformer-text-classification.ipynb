# Transformer — Text Classification

Implementasi Transformer Encoder untuk klasifikasi teks (NLP).

```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
from src.models.transformer import TransformerClassifier
```

```python
vocab = {'<pad>': 0, 'good': 1, 'bad': 2, 'great': 3, 'terrible': 4, 'amazing': 5, 'awful': 6}
texts = [
    [1, 3, 5],
    [2, 4, 6],
    [3, 5],
    [4, 6],
    [1, 5],
    [2, 6],
]
labels = [1, 0, 1, 0, 1, 0]
max_len = max(len(t) for t in texts)
X = torch.tensor([t + [0] * (max_len - len(t)) for t in texts])
y = torch.tensor(labels)
```

```python
model = TransformerClassifier(vocab_size=len(vocab), d_model=32, nhead=2, num_layers=2, num_classes=2)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)
```

```python
for epoch in range(20):
    optimizer.zero_grad()
    loss = criterion(model(X), y)
    loss.backward()
    optimizer.step()
    print(f'Epoch {epoch+1}, Loss: {loss.item():.4f}')
```

```python
preds = model(X).argmax(dim=1)
print(f'Predictions: {preds.tolist()}')
print(f'Ground Truth: {y.tolist()}')
print(f'Accuracy: {(preds == y).sum().item() / len(y) * 100:.0f}%')
```
